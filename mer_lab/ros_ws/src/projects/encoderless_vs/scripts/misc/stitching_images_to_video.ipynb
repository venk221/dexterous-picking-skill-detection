```python
import cv2
import numpy as np

  # choose codec according to format needed
fourcc = cv2.VideoWriter_fourcc('X', 'V', 'I', 'D') 
# fourcc = cv2.VideoWriter_fourcc('DIVX') 

video=cv2.VideoWriter('video.avi', fourcc, 10,(720,1280))

for i in range(0,88):
    img = cv2.imread('/home/jani/Documents/pics/CV-Depth' + str(i)+'.tiff')
    video.write(img)
cv2.destroyAllWindows()
video.release()
```

```python
import cv2
import glob
import numpy as np

image_list = glob.glob('/home/jani/Documents/original_depth_image/*.jpg')
image_list.sort()
#print(image_list)

codec = cv2.VideoWriter_fourcc('X', 'V', 'I', 'D')
video = cv2.VideoWriter('original.avi', codec, 10, (300, 300))

for img_name in image_list:
    img = cv2.imread(img_name)
    img = cv2.resize(img, (300, 300))
    video.write(img)

video.release()
```
