```python
import matplotlib.pyplot as plt
import csv
import numpy as np
  
# x1 = []
# x2 = []
# x3 = []
y1 = []
y2 = []
y3 = []
  
with open('coeff_record.csv','r') as csvfile:
    plots = csv.reader(csvfile, delimiter = ',')
    list_data = []
    for row in plots:
        list_data.append(row)
      
    for row in list_data[1:]:
#         print(row)
        y1.append(float(row[0]))
        y2.append(float(row[1]))
        y3.append(float(row[2]))

x = np.linspace(1,90, 90).tolist()
print(len(x))
print(type(x))

print(y1)
print(y2)
print(y3)
```

```python
fig = plt.figure()
fig, ax = plt.subplots()
# ax = fig.add_subplot(111)
# fig.show()
# fig.canvas.draw()
# ax.clear()
# ax.set_xlim([0, 140])
# ax.set_ylim([0, 1])
# ax.set_ylim([0, 60])
# ax.set_ylim([0, 60])

#ax.set_ylim([0,300])
ax.plot(x, y1, 'b')
ax.plot(x, y2, 'g')
ax.plot(x, y3, 'r')

ax.set_ylabel('coeffs')
ax.set_xlabel('frames')

# ax.set_yticklabels([])
# ax.set_xticklabels([])
# ax.set_yticks(y1)
# ax.set_yticks(y2, minor = True)
# ax.set_yticks(y3)

# ax.plot(x, y1)
# ax.plot(x, y2)
# ax.plot(x, y3)
ax.legend(['coeff1','coeff2', 'coeff3'])
plt.show()
# ax.legend(['coeff1','coeff2', 'coeff3'])
```
