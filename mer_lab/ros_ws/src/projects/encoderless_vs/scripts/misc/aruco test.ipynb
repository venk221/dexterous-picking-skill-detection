```python
import cv2
import cv2.aruco as aruco
```

```python
image = cv2.imread('/home/jani/Documents/pics/CV-Depth150.tiff') 

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) 

cv2.imwrite('aruco_gray.jpg', gray) 

aruco_dict = aruco.Dictionary_get(aruco.DICT_4X4_50) 

arucoParameters = aruco.DetectorParameters_create() 

corners, ids, rejectedImgPoints = aruco.detectMarkers(gray, aruco_dict, parameters=arucoParameters) 

print(ids) 
print(corners)

frame = aruco.drawDetectedMarkers(image, corners) 

# rej_frame = aruco.drawDetectedMarkers(image, rejectedImgPoints, borderColor=(100, 0, 240)) 

# cv2.imwrite('aruco_display.jpg', frame) 
```
