```python
# # get the path/directory
# folder_dir = 
# # files = sorted(os.listdir(parent_folder), reverse=True)
file_list = []
# id_list = []
# image_list = []
kp_list = []

for files in os.listdir(folder_dir): 
    # check if the file ends with jpg
#     if (files.endswith(".rgb.jpg")):
#         id = int(files.split('.')[0])
#         id_list.append(id)
#         image_file = str(files)
#         image_list.append(image_file)
    if (files.endswith(".json")):
        file_list.append(os.path.join(folder_dir, files))

print(file_list[0])

for file in file_list:
    with open(file, encoding='utf-8', mode='r') as f:
        print(type(file))
        print(file)
        data = json.load(f)
        id = int(file.split('/')[-1].split('.')[0])
        image_rgb = str(id)+".jpg"
#         image_dep = str(id)+".dep.jpg"
        kp1=data["objects"][0]["keypoints"][0]["projected_location"]
        kp1.append(1)
        kp2=data["objects"][0]["keypoints"][1]["projected_location"]
        kp2.append(1)
        kp3=data["objects"][0]["keypoints"][2]["projected_location"]
        kp3.append(1)
        kp4=data["objects"][0]["keypoints"][3]["projected_location"]
        kp4.append(1)
        kp5=data["objects"][0]["keypoints"][4]["projected_location"]
        kp5.append(1)
        kp6=data["objects"][0]["keypoints"][5]["projected_location"]
        kp6.append(1)
        bbox1=[kp1[0]-1,kp1[1]-1, kp1[0]+1, kp1[1]+1]
        bbox2=[kp2[0]-1,kp2[1]-1, kp2[0]+1, kp2[1]+1]
        bbox3=[kp3[0]-1,kp3[1]-1, kp3[0]+1, kp3[1]+1]
        bbox4=[kp4[0]-1,kp4[1]-1, kp4[0]+1, kp4[1]+1]
        bbox5=[kp5[0]-1,kp5[1]-1, kp5[0]+1, kp5[1]+1]
        bbox6=[kp6[0]-1,kp6[1]-1, kp6[0]+1, kp6[1]+1]
        
        new_data = {"id": id, "image_rgb": image_rgb,"bboxes":[bbox1,bbox2,bbox3,bbox4,bbox5,bbox6], "keypoints": [[kp1], [kp2], [kp3], [kp4], [kp5], [kp6]]}
        json_obj = json.dumps(new_data, indent=4)
        filename =  parent_path + "annotations/"+str(id)+".json"
        with open(filename, "w") as outfile:
                outfile.write(json_obj)
```
