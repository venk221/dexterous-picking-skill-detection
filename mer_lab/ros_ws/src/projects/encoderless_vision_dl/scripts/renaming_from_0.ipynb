```python
import os
import numpy as np
import json
from os.path import expanduser
import cv2
```

```python
home = expanduser("~")

parent_path =  home + "/Pictures/" + "Data/"

folder_dir = parent_path + "merge_folder-2022-11-04-13-26"+ "/"
```

```python
count = 0
# count increase by 1 in each iteration
# iterate all files from a directory
for file_name in sorted(os.listdir(folder_dir)):
    print(file_name)
    if file_name.endswith('jpg'):
    # Construct old file name
        source = folder_dir + file_name

    # Adding the count to the new file name and extension
        destination = folder_dir + str(count) + ".jpg"

    # Renaming the file
        os.rename(source, destination)
        count = count + 1

count = 0

for file_name in sorted(os.listdir(folder_dir)):
    if file_name.endswith('json'):
    # Construct old file name
        source = folder_dir + file_name

    # Adding the count to the new file name and extension
        destination = folder_dir + str(count) + ".json"

    # Renaming the file
        os.rename(source, destination)
        count = count + 1
```
